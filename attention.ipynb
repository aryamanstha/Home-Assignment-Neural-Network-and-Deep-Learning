```python
import numpy as np
```

```python
Q = np.array([[1, 0, 1, 0], [0, 1, 0, 1]])
K = np.array([[1, 0, 1, 0], [0, 1, 0, 1]])
V = np.array([[1, 2, 3, 4], [5, 6, 7, 8]])
```

```python
def scaled_dot_product_attention(Q, K, V):

    # Compute the dot product of Q and K transpose
    scores = Q @ K.T

    # Scale the scores by the square root of the dimension of K
    d_k = K.shape[-1]
    scaled_scores = scores / np.sqrt(d_k)

    # Compute the attention weights using softmax
    exp_scores = np.exp(
        scaled_scores - np.max(scaled_scores, axis=-1, keepdims=True)
    )

    # Normalize the attention weights
    attention_weights = exp_scores / np.sum(exp_scores, axis=-1, keepdims=True)

    # Compute the output by multiplying the attention weights with V
    output = attention_weights @ V

    return attention_weights, output
```

```python
weights, output = scaled_dot_product_attention(Q, K, V)

print("Attention Weights:")
print(weights)

print("\nFinal Output:")
print(output)
```
